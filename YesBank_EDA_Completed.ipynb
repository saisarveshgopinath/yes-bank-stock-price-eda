## 3. ***Data Wrangling***

### Data Wrangling Code

````python
required = {"Date", "Open", "High", "Low", "Close"}if required.difference(df.columns):    raise ValueError(f"Missing columns: {required.difference(df.columns)}")df = df.copy()df["Date"] = pd.to_datetime(df["Date"], format="%b-%y", errors="coerce")for c in ["Open","High","Low","Close"]:    df[c] = pd.to_numeric(df[c], errors="coerce")if df[["Date","Open","High","Low","Close"]].isna().any().any():    raise ValueError("Invalid or missing date/price values.")df = df.sort_values("Date").drop_duplicates().reset_index(drop=True)df["Year"] = df["Date"].dt.yeardf["Monthly_Return_Pct"] = df["Close"].pct_change()*100df["High_Low_Range_Pct"] = (df["High"]-df["Low"])/df["Close"]*100df["Running_Peak"] = df["Close"].cummax()df["Drawdown_Pct"] = (df["Close"]/df["Running_Peak"]-1)*100assert (df["High"] >= df[["Open","Close","Low"]].max(axis=1)).all()assert (df["Low"] <= df[["Open","Close","High"]].min(axis=1)).all()print(df.shape, df["Date"].min(), df["Date"].max())
````

### What all manipulations have you done and insights you found?

Dates and prices were parsed to proper types, rows sorted, and OHLC high/low consistency checked. No missing or duplicate rows needed imputation or removal. Added year, monthly Close return, high-low range as a percentage of close, running peak, and drawdown. The first monthly return is undefined because no prior month is provided. Extreme observations were retained and not assigned unsupported causes.
